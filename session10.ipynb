```python
#task3 Product analysis
df.groupby('Category')['Revenue'].sum().idxmax()
#Sir laptop generated the highest revenue of 610000
df.groupby('Category')['Quantity'].sum()
#sir, Phone sold at the highest quantity {
#Laptop    13
#Phone     15
#Tablet    18}
df.groupby('Category')['Rating'].mean()
#Laptop has the highest average rating category with 4.8 avg rating i.e., more tha phone an tablet rating
#4.18 as both respectively
```

```python
#task4 city analysis
df.groupby('City')['Revenue'].sum().idxmax()
#Delhi generates the highest revenue of 435000,we might make our serives more flexible in delhi to increase sales
df.groupby('City')['Quantity'].sum().idxmax()
#Delhi bought the highest quantity , makes sense why the revenue of delhi is highest
df.groupby('City')['Revenue'].mean()
#City    |  average
#Delhi      54375.000000
#Lucknow    42666.666667
#Mumbai     57600.000000
#Pune       35750.000000
df.groupby('City')['Order_ID'].count()
#City    | orders 
#Delhi      8
#Lucknow    3
#Mumbai     5
#Pune       4
#overall , Delhi for our company is milestone for grwoth but , it tells us that 
#we have to improve our services and marketing in other cities 
```

```python
#task5 Customer analysis
df.groupby('Customer')['Revenue'].sum()
#Ali contributes the highest revenue , hence , he is the highest spending customer
df.groupby('Customer')['Order_ID'].count().idxmax()
#Ali has the total of 4 orders , he placed most of the orders from our company
df.groupby('Customer')['Revenue'].mean()
#most of the customer on average spend bw (522000 to 56000) ,
#so the product liwes bw this range has more sale 

#We should give ALI offer for his next order
# we must focus on the quality of product liwes in prices bw 52000 to 56000
```

```python
#task 6 Payment analysis
df.groupby('Payment')['Revenue'].sum().idxmax()
#UPI method generates highest total revenue
df.groupby('Payment')['Order_ID'].count()
#total order placed by cash and card = 6 and UPI = 8
df.groupby('Payment')['Rating'].mean()
#Avg ratig for card and cash= 4.3 whie of UPI = 4.5

#The UPI payment method seems strong and has a big hand in generating hightes revenue and excellent rating 
```
